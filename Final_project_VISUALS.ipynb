```python
# Univariate Analysis
```

```python
df['Climate_Risk_Score'].describe() # Statistical Description of Climate_Risk_Score
```

```python
#The total count of data_value is 4999

#The central tendency :
    # Mean = 57.597
    # Median (50%) = 58
#Since mean > median, the data is right-skewed

#Standard deviation = 22.030
```

```python
print('variance:', df['Climate_Risk_Score'].var())            #variance
print('standard deviation:', df['Climate_Risk_Score'].std())  #standard deviation
```

```python
# variance-Measures the average squared deviation from the mean
# standard deviation- Standard Deviation is a statistical measure that tells you how spread out the values in your data are from the mean
```

```python
print('Skewness:',df['Climate_Risk_Score'].skew())  #Skewness
print('Kurtosis:',df['Climate_Risk_Score'].kurt())  #Kurtosis
```
