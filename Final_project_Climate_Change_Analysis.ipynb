```python
# BAR PLOT - POPULATION AFFECTED
```

```python
top_countries = df.groupby('Country')['Population_Affected_m'].mean().sort_values(ascending=False).head(10)

top_countries.plot(kind='bar')
plt.xlabel("Country")
plt.ylabel("Population Affected (millions)")
plt.title("Top Countries by Population Affected")
plt.show()
```

```python
# This bar chart shows countries with the highest affected population.
# X-axis: Country  Y-axis: Population affected
# Key Findings- # Indian population is mostly affected
                # Few countries experience severe human impact.
                # Indicates higher vulnerability in those regions.
```

```python
#LINE PLOT – FLOOD RISK VS AVERAGE POPULATION AFFECTED
```

```python
plt.figure()
flood_pop = df.groupby('Flood_Risk')['Population_Affected_m'].mean()
flood_pop.plot(kind='line', marker='o')
plt.xlabel("Flood Risk Category")
plt.ylabel("Average Population Affected (millions)")
plt.title("Flood Risk vs Population Affected")
plt.show()
```

```python
# This line plot shows how population impact changes with different flood risk levels.
# X-axis: Flood Risk Category (Low, Medium, High)  Y-axis: Average Population Affected (in millions)
#An upward trend indicates that higher flood risk leads to greater population impact. A flat line would indicate no relationship.
# Key Findings- # Higher flood risk categories are associated with larger affected populations.
                # This suggests flood risk is a major factor influencing human impact.
                # Indicates need for better flood management in high-risk areas.
```

```python
# VIOLIN PLOT – REGION VS CLIMATE RISK SCORE
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(x='Region', y='Temperature_Change_C', data=df)
plt.xticks(rotation=45)
plt.xlabel("Region")
plt.ylabel("Temperature Change (°C)")
plt.title("Region vs Temperature Change")
plt.show()
```

```python
# This violin plot shows the distribution of climate risk scores across different regions
# X-axis: Region Y-axis: Climate Risk Score
# It combines a boxplot and a density plot to display how climate risk scores are distributed in each region.

# Wider sections indicate a higher concentration of regions with similar risk scores.
# Narrow sections indicate fewer regions in that range.
# The shape of the violin shows the spread and distribution pattern of climate risk.

# Key Findings
# Some regions have a wider spread, indicating high variability in climate risk.
# Regions with wider upper sections have more high-risk areas.
# Uneven shapes across regions indicate that climate risk is not uniformly distributed.
# Certain regions show consistently higher risk levels, highlighting climate vulnerability.
```

```python
# CORRELATION HEAT MAP
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# This is a correlation matrix heatmap showing the relationship between different climate variables.
#Values range from -1 to +1:
#+1 → Strong positive correlation
#0 → No correlation
#-1 → Strong negative correlation

# Key Findings

#Strong Correlation
#Extreme_Index vs Wildfire_Incidents ≈ 0.98
#Very strong positive relationship
#As extreme climate events increase, wildfire incidents also increase significantly

#High_Risk ↔ Climate_Risk_Score ≈ 0.82
#Strong positive correlation
#High-risk classification is strongly dependent on the climate risk score

# Weak / No Correlation (Most Variables)
#Most values are very close to 0 (e.g., 0.01, -0.02, 0.005), which means

#CO2_Emissions vs Temperature_Change_C ≈ 0.021
#Sea_Level_Rise vs Temperature_Change ≈ 0.0057
#Rainfall_Change vs others ≈ near 0
```
